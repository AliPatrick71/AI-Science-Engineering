**Objective:** Write a Python program that converts temperatures from Celsius to Fahrenheit, compares temperatures against a threshold, and uses functions and loops.

```python

# Week 2 Assignment: Temperature Conversion

# This function converts Celsius to Fahrenheit.
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit


# Set the temperature threshold in Celsius.
threshold = 25

# Ask the user to enter a temperature.
temperature = float(input("Enter a temperature in Celsius: "))

# Convert the temperature and display the result.
fahrenheit = celsius_to_fahrenheit(temperature)
print(f"{temperature:.1f}°C is equal to {fahrenheit:.1f}°F.")

# Check whether the temperature is above or below the threshold.
if temperature > threshold:
    print(f"The temperature is above {threshold}°C.")
elif temperature < threshold:
    print(f"The temperature is below {threshold}°C.")
else:
    print(f"The temperature is exactly {threshold}°C.")


# Ask the user to enter several temperatures separated by commas.
temperatures_input = input(
    "Enter several temperatures in Celsius, separated by commas: "
)

# Convert the input into a list of numbers.
temperatures = [
    float(value.strip()) for value in temperatures_input.split(",")
]

# Use a loop to convert and display each temperature.
print("\nTemperature Conversion Results:")

for temp in temperatures:
    converted = celsius_to_fahrenheit(temp)
    print(f"{temp:.1f}°C = {converted:.1f}°F")
```
